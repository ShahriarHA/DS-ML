```python
from ucimlrepo import fetch_ucirepo 

covertype = fetch_ucirepo(id=31) # fetch dataset

x = covertype.data.features #features
y = covertype.data.targets  #class
```

```python
x.head()
```

```python
y.head()
```

```python
print(x.shape,y.shape)
```

```python
from sklearn.model_selection import train_test_split
```

```python
# train test split
xtrain, xtest, ytrain, ytest = train_test_split(x, y, random_state=42, test_size=.30)
```

```python
print(xtrain.shape,ytrain.shape)
print(xtest.shape,ytest.shape)
```

```python
# test and validation
x_test, x_val, y_test, y_val = train_test_split(xtest, ytest, random_state=42, test_size=.50)
```

```python
print(x_test.shape,y_test.shape)
print(x_val.shape,y_val.shape)
```

```python
import xgboost as xgb
from catboost import CatBoostClassifier
from lightgbm import LGBMClassifier

from sklearn.metrics import accuracy_score, classification_report
```

# XGBoost

```python
xgboost  = xgb.XGBClassifier(eval_metric='logloss', random_state=42)
xgboost.fit(xtrain,ytrain)
```

```python
print(ytrain.shape,y_test.shape)
```

```python
print("-----ytrain value count------")
print(ytrain.value_counts())
print("-----y_test value count------")
print(y_test.value_counts())
```

```python
y.value_counts()
```

```python
y_new = y-1
y_new.value_counts()
```

```python
# train test split
xtrain1, xtest1, ytrain1, ytest1 = train_test_split(x, y_new, random_state=42, test_size=.30)
```

```python
print(xtrain1.shape,ytrain1.shape)
print(xtest1.shape,ytest1.shape)
print("-----ytrain1 value count------")
print(ytrain1.value_counts())
print("-----ytest1 value count------")
print(ytest1.value_counts())
```

```python
# test and validation
x_test1, x_val1, y_test1, y_val1 = train_test_split(xtest1, ytest1, random_state=42, test_size=.50)
```

```python
print(x_test1.shape,y_test1.shape)
print(x_val1.shape,y_val1.shape)
print("-----y_test1 value count------")
print(y_test1.value_counts())
print("-----y_val1 value count------")
print(y_val1.value_counts())
```

```python
xgboost1 = xgb.XGBClassifier(eval_metric='logloss', random_state=42)
xgboost1.fit(xtrain1,ytrain1)
```

```python
xgboost1_y_predict = xgboost1.predict(x_test1)
```

```python
accuracy_score(y_test1,xgboost1_y_predict)
```

```python
xgboost1.score(xtrain1,ytrain1)
```

# CatBoost

```python
cat_model = CatBoostClassifier(random_state=42)
cat_model.fit(xtrain1,ytrain1)
```

```python
catModel_y_pre = cat_model.predict(x_test1)
```

```python
accuracy_score(y_test1,catModel_y_pre)
```

```python
cat_model.score(xtrain1,ytrain1)
```
