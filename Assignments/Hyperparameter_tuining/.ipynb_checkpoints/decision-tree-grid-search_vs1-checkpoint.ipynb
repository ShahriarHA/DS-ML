# Train Test Split

```python
from sklearn.model_selection import train_test_split
```

```python
xtrain,xtest,ytrain,ytest = train_test_split(x,y,test_size=0.30,random_state=42)
```

```python
print(f'shape of xtrain = {xtrain.shape},\nshape of ytrain{ytrain.shape}')
```

```python
print(f'shape of xtest = {xtest.shape},\nshape of ytest{ytest.shape}')
```

# test and validation

```python
x_test,x_val,y_test,y_val = train_test_split(xtest,ytest,test_size=0.50,random_state=42)
```

```python
print(f'shape of x_test = {x_test.shape},\nshape of x_val{y_test.shape}')
```

```python
print(f'shape of x_val = {x_val.shape},\nshape of y_val{y_val.shape}')
```

```python
from sklearn import tree
from sklearn.tree import DecisionTreeClassifier
from sklearn.ensemble import RandomForestClassifier

from sklearn.metrics import accuracy_score, classification_report
```

# Decision Tree model

```python
dt_model = DecisionTreeClassifier(random_state=1)
```

```python
dt_model.fit(xtrain,ytrain)
```

```python
dt_y_predict = dt_model.predict(x_test)
```

```python
dt_y_predict
```

```python
len(dt_y_predict)
```

## Evaluate decision tree model

```python
print(f'Accuracy of DT: {accuracy_score(y_test,dt_y_predict)}')
print("----------Classification report of DT---------")
print(classification_report(y_test,dt_y_predict))
```

# Random Forest model

```python
rf_model = RandomForestClassifier(random_state=42)
```

```python
rf_model.fit(xtrain,ytrain)
```

```python
rf_y_predict = rf_model.predict(x_test)
```

```python
rf_y_predict
```

```python
len(rf_y_predict)
```

## Evaluate Random Forest model

```python
print(f'Accuracy of Random Forest Model: {accuracy_score(y_test,rf_y_predict)}')
print("----------Classification report of Random Forest---------")
print(classification_report(y_test,rf_y_predict))
```

# Pre - Pruning
## initializing hyperperameters for both DT & RF

```python
dt_params = {
    'criterion': ['gini', 'entropy'],
    'splitter': ['best', 'random'],
    'max_depth': [10, 15, 20, 25, 30, 40, None],
    'min_samples_split': [2, 5, 10, 20, 50],
    'min_samples_leaf': [1, 5, 10, 15, 20]
}
```

# perform Grid search for Decision tree

```python
from sklearn.model_selection import GridSearchCV, RandomizedSearchCV
```

```python
dt_grid_search = GridSearchCV(estimator = DecisionTreeClassifier(random_state=42),
                              param_grid = dt_params, cv=5)
dt_grid_search.fit(xtrain,ytrain)
```

```python
dt_grid_search_tuing_result = pd.DataFrame(dt_grid_search.cv_results_)
```

```python
dt_grid_search_tuing_result.head()
```

```python
dt_grid_search_tuing_result.shape
```

```python
dt_grid_search_tuing_result.sort_values('rank_test_score').head(5)
```

```python
dt_grid_search.best_params_
```

```python
dt_gs_best_model = dt_grid_search.best_estimator_
```

```python
# validation score
dt_val_score_gs = dt_gs_best_model.score(x_val,y_val)
dt_val_score_gs
```

```python
# test score
dt_test_score_gs = dt_gs_best_model.score(x_test,y_test)
dt_test_score_gs
```

```python
y_pre_dt_gs = dt_grid_search.predict(x_test)
y_pre_dt_bestMoel_gs = dt_gs_best_model.predict(x_test)
print(y_pre_dt_gs)
print(y_pre_dt_bestMoel_gs)
```

```python
print("=========> 'Accuracy and classification report <============'")
print(f"Accuracy 1: {accuracy_score(y_test,y_pre_dt_gs)}")
print(f"Accuracy 2: {accuracy_score(y_test,y_pre_dt_bestMoel_gs)}")

print("---------- Classification report 1----------")
print(classification_report(y_test,y_pre_dt_gs))

print("---------- Classification report 2----------")
print(classification_report(y_test,y_pre_dt_bestMoel_gs))
```
