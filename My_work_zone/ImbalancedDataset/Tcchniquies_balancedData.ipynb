```python
from sklearn.model_selection import train_test_split
```

```python
x = df.drop("Churn",axis=1)
```

```python
y = df["Churn"]
```

```python
y.head()
```

```python
print(x.shape,y.shape)
```

```python
from sklearn.preprocessing import LabelEncoder
```

```python
le = LabelEncoder()
```

```python
xtrain, xtest, ytrain, ytest = train_test_split(x,y, test_size=0.30, random_state=42)
```

```python
print(xtrain.shape,ytrain.shape)
print(xtest.shape,ytest.shape)
```

```python
xtrain.head()
```

```python
ytrain.head()
```

```python
ytest.head()
```

```python
ytrain_preprocessed = le.fit_transform(ytrain)
ytest_preprocessed = le.transform(ytest)
```

```python
ytrain_preprocessed
```

```python
ytest_preprocessed
```

```python
df.columns
```

```python
from pandas.core.dtypes.common import is_numeric_dtype
```

```python
categorical_columns = [col for col in xtrain.columns if not is_numeric_dtype(xtrain[col])]
numeric_columns = [col for col in xtrain.columns if is_numeric_dtype(xtrain[col])]
```

```python
print(categorical_columns)
print(len(categorical_columns))
```

```python
print(numeric_columns)
print(len(numeric_columns))
```

```python
from sklearn.compose import ColumnTransformer
from sklearn.preprocessing import OneHotEncoder
```

```python
pre_processing = ColumnTransformer(
    transformers=[
        ("cat",OneHotEncoder(handle_unknown="ignore"),categorical_columns)
    ]
)
```

```python
from sklearn.pipeline import Pipeline
from sklearn.linear_model import LogisticRegression
from sklearn.tree import DecisionTreeClassifier
```

```python
model = Pipeline(steps=[
    ("Pre-processing_steps",pre_processing),
    ("Classifier1",LogisticRegression()),
    # ("Classifier2",DecisionTreeClassifier())

])
```

```python
model.fit(xtrain,ytrain_preprocessed)
```

```python
model.predict(xtest)
```

```python
model.score(xtest,ytest_preprocessed)
```

```python
pre_processing1 = ColumnTransformer(
    transformers=[
        ("cat",OneHotEncoder(handle_unknown="ignore"),categorical_columns)
    ]
)
```

```python
model2 = Pipeline(steps=[
    ("Pre-processing_steps",pre_processing1),
    # ("Classifier1",LogisticRegression()),
    ("Classifier2",DecisionTreeClassifier())

])
```

```python
model2.fit(xtrain,ytrain_preprocessed)
```

```python
model2.predict(xtest)
```

```python
model2.score(xtest,ytest_preprocessed)
```
